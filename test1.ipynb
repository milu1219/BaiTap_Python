```python
import secrets
import string

class PasswordGenerator:
    def __init__(self,length,use_upper,use_lower,use_digits,use_symbols):# Khởi tạo với init
        self._length = length# 
        self._use_upper = use_upper
        self._use_lower = use_lower
        self._use_digits = use_digits
        self._use_symbols = use_symbols
    def GeneratePassword(self)->str:
        if self._use_upper:
            char_pool += string.ascii_uppercase
        if self._use_lower:
            char_pool += string.ascii_lowercase
        if self._use_digits:
            char_pool += string.digits
        if self._use_symbols:
            char_pool += "!@#$%^&*()_+-=[]{}|;:,.<>?"

        if not char_pool:
            print("You need to enter at least one character!")
        
        return "".join(secrets.choice(char_pool) for _ in range(self.length)) # bốc đại 1 chữ, dán dính chữ lại rồi bắt đầu lặp n lần với độ dài vừa nhập
            
```
