```python
import secrets
import string

class PasswordGenerator:
    def __init__(self,length,use_uppercase,use_lowercase,use_numbers,use_special):# Khởi tạo với init
        self._length = length# 
        self._use_uppercase = use_uppercase
        self._use_lowercase = use_lowercase
        self._use_numbers = use_numbers
        self._use_special = use_special
    def GeneratePassword(self)->str:
        if self._use_uppercase:
            char_pool += string.ascii_uppercase
        if self._use_lowercase:
            char_pool += string.ascii_lowercase
        if self._use_numbers:
            char_pool += string.digits
        if self._use_special:
            char_pool += "!@#$%^&*()_+-=[]{}|;:,.<>?"

        if not char_pool:
            print("You need to enter at least one character!")
        
        return "".join(secrets.choice(char_pool) for _ in range(self.length)) # bốc đại 1 chữ, dán dính chữ lại rồi bắt đầu lặp n lần với độ dài vừa nhập
            
```
